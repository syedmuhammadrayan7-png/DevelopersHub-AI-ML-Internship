# Task 1: Exploring and Visualizing Iris Dataset

## Objective
Learn how to load, inspect, analyze, and visualize a dataset using Python libraries.

```python
# Importing required libraries

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load iris dataset

df = sns.load_dataset("iris")
```

```python
df.head()
```

```python
# Show rows and columns

df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
# Scatter plot

sns.scatterplot(
    x='sepal_length',
    y='sepal_width',
    hue='species',
    data=df
)

plt.title("Sepal Length vs Sepal Width")
plt.show()
```

```python
# Histogram

df.hist(figsize=(10,8))

plt.show()
```

```python
# Box plot

sns.boxplot(data=df)

plt.title("Boxplot of Iris Features")
plt.show()
```

## Observations and Insights

1. The dataset contains 150 rows and 5 columns.

2. There are three flower species:
- Setosa
- Versicolor
- Virginica

3. No missing values were found in the dataset.

4. Scatter plots show relationships between flower measurements.

5. Histograms help understand the distribution of feature values.

6. Box plots help identify potential outliers.

7. Different flower species show different measurement patterns.
